```python
job_data = {
    "Job_ID": [1, 2, 3, 4, 5],
    "Role": ["Data Analyst", "Data Scientist", "Business Analyst", "Data Engineer", "ML Engineer"],
    "Company": ["ABC Corp", "XyZ Ltd", "DataWorks", "InfraTech", "AI Labs"],
    "Skills": [
        "Excel SQL Python",
        "Python ML Statistics",
        "Excel SQL PowerBI",
        "Python SQL Spark",
        "Python ML DeepLearning"
    ]
}

print(job_data)
```

```python
print("\n=== Job Postings ===")

for i in range(len(job_data["Job_ID"])):
    print(
        f"ID: {job_data['Job_ID'][i]} | "
        f"Role: {job_data['Role'][i]} | "
        f"Company: {job_data['Company'][i]} | "
        f"Skills: {job_data['Skills'][i]}"
    )
```

```python
num_jobs = int(input("How many job postings do you want to add? "))

for i in range(num_jobs):

    role = input("Enter role: ")
    company = input("Enter company: ")
    skills = input("Enter skills (space-separated or comma-separated): ")

    new_id = max(job_data["Job_ID"]) + 1

    job_data["Job_ID"].append(new_id)
    job_data["Role"].append(role)
    job_data["Company"].append(company)
    job_data["Skills"].append(skills)

print("new postings added successfully!")
```

```python
import re

def clean_skills(skill_text):

    skill_text = skill_text.lower()

    skill_text = skill_text.replace(",", " ")

    skill_text = re.sub(r"[^\w\s]", "", skill_text)

    skill_text = " ".join(skill_text.split())

    return skill_text


job_data["Skills"] = [
    clean_skills(skill)
    for skill in job_data["Skills"]
]

print(" Cleaned Skills :")

for skill in job_data["Skills"]:
    print(skill)
```

```python
all_skills = []
for skill_string in job_data["Skills"]:
    skills = skill_string.split()
    all_skills.extend(skills)

skill_count = {}
for skill in all_skills:
    if skill in skill_count:
        skill_count[skill] += 1
    else:
        skill_count[skill] = 1

sorted_skills = sorted(
    skill_count.items(),
    key=lambda x: x[1],
    reverse=True
)

print(" Top Skills by Frequency ")
top_n = min(5, len(sorted_skills))
for skill, count in sorted_skills[:top_n]:
    print(f"{skill}: {count}")
```

```python
total_jobs = len(job_data["Job_ID"])
unique_skills = len(set(all_skills))
python_jobs = []
for i in range(len(job_data["Job_ID"])):
    if "python" in job_data["Skills"][i].split():
        python_jobs.append(job_data["Role"][i])

print("Statistics")
print("Total Jobs:", total_jobs)
print("Unique Skills:", unique_skills)
print("Jobs mentioning Python:", python_jobs)
```

```python
def find_jobs_by_skill(skill_query):

    skill_query = skill_query.lower().strip()

    matches = []

    for i in range(len(job_data["Job_ID"])):

        skills = job_data["Skills"][i].split()

        if skill_query in skills:

            matches.append(
                (
                    job_data["Job_ID"][i],
                    job_data["Role"][i],
                    job_data["Company"][i]
                )
            )

    return matches


print(" Search Result ")
print(find_jobs_by_skill("sql"))
```

```python
common_roles = list(filter(
    lambda role_index:
    any(skill_count.get(skill, 0) >= 2
        for skill in job_data["Skills"][role_index].split()),
    range(len(job_data["Role"]))
))

common_roles = [
    job_data["Role"][i]
    for i in common_roles
]

print("\nRoles with at least one common skill:")

print(common_roles)
```

```python
unique_skill_set = sorted(set(all_skills))

print(" Unique Skills ")

print(unique_skill_set)
```

```python
import csv

with open("job_skills.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    # Write the header
    writer.writerow(["Job_ID", "Role", "Company", "Skills"])

    # Write the rows
    for i in range(len(job_data["Job_ID"])):
        writer.writerow([
            job_data["Job_ID"][i],
            job_data["Role"][i],
            job_data["Company"][i],
            job_data["Skills"][i]
        ])
```
